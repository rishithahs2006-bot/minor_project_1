```python
import numpy as np
from datetime import datetime
```

```python
with open("hostel_bois.txt", "r", encoding="utf-8") as file:
    content = file.read()

print("File loaded successfully!")
print(content[:500])
```

```python
messages = []

lines = content.split("\n")

for line in lines:
    if " - " in line and ": " in line:
        try:
            date_time, message = line.split(" - ", 1)
            sender, text = message.split(": ", 1)

            messages.append({
                "datetime": date_time,
                "sender": sender,
                "message": text
            })
        except:
            pass

print("Total messages:", len(messages))
print(messages[:5])
```

```python
person_count = {}

for msg in messages:
    sender = msg["sender"]

    if sender in person_count:
        person_count[sender] += 1
    else:
        person_count[sender] = 1

print("MESSAGES PER PERSON")
print("-" * 30)

for person, count in person_count.items():
    print(person, ":", count)
```

```python
day_count = {}
hour_count = {}

for msg in messages:
    try:
        dt = datetime.strptime(msg["datetime"], "%d/%m/%y, %H:%M")

        day = dt.strftime("%d/%m/%Y")
        hour = dt.hour

        day_count[day] = day_count.get(day, 0) + 1
        hour_count[hour] = hour_count.get(hour, 0) + 1

    except:
        pass

busiest_day = max(day_count, key=day_count.get)
busy_hour = max(hour_count, key=hour_count.get)

print("BUSIEST DAY:", busiest_day, "-", day_count[busiest_day], "messages")
print("BUSIEST HOUR:", f"{busy_hour:02d}:00 - {busy_hour+1:02d}:00")
```

```python
# Busiest day and busiest hour

day_count = {}
hour_count = {}

for msg in messages:
    try:
        dt = datetime.strptime(msg["datetime"], "%d/%m/%y, %H:%M")

        day = dt.strftime("%d/%m/%Y")
        hour = dt.hour

        day_count[day] = day_count.get(day, 0) + 1
        hour_count[hour] = hour_count.get(hour, 0) + 1

    except Exception:
        pass

print("BUSIEST DAY:")
print(max(day_count, key=day_count.get), "-", max(day_count.values()), "messages")

print()

busy_hour = max(hour_count, key=hour_count.get)
print("BUSIEST HOUR:")
print(f"{busy_hour:02d}:00 - {busy_hour+1:02d}:00")
```

```python
# Most Active User

most_active = max(person_count, key=person_count.get)

print("MOST ACTIVE USER")
print("----------------------------")
print("Name:", most_active)
print("Messages:", person_count[most_active])
```

```python
print("PROJECT SUMMARY")
print("------------------------")
print("Total Messages :", len(messages))
print("Most Active User :", most_active)
print("Messages Sent :", person_count[most_active])
print("Busiest Day :", busiest_day)
print("Busiest Hour :", f"{busy_hour:02d}:00 - {busy_hour+1:02d}:00")
```

```python
import matplotlib.pyplot as plt

names = list(person_count.keys())
counts = list(person_count.values())

plt.figure(figsize=(8,5))
plt.bar(names, counts)
plt.title("Messages per Person")
plt.xlabel("Users")
plt.ylabel("Number of Messages")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(7,7))
plt.pie(counts, labels=names, autopct="%1.1f%%")
plt.title("Message Distribution Among Users")

plt.show()
```
